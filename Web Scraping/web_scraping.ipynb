# Extracting the features of all books

````python
titles=[]
links=[]
prices=[]
avaliability=[]
rating=[]
for i in range(1,51):
    url=f"https://books.toscrape.com/catalogue/page-{i}.html"
    response=requests.get(url)
    soup=BeautifulSoup(response.text)
    books=soup.find_all("article",class_="product_pod")
    for book in books:
        title=book.find("h3").find("a")["title"]
        titles.append(title)
        link=book.find("h3").find("a")["href"]
        links.append(link)
        price=book.find("p",class_="price_color").text
        prices.append(price)
        avaliable=book.find("p",class_="instock availability").text.strip()
        avaliability.append(avaliable)
        star=book.find("p",class_="star-rating")["class"][1]
        rating.append(star)
````

````python
import pandas as pd

# Create DataFrame
df = pd.DataFrame({
    "Title": titles,
    "Link": links,
    "Price": prices,
    "Availability": avaliability,
    "Rating": rating
})
df.head()
````

````python
df.to_csv("books_data.csv",index=False)
````
