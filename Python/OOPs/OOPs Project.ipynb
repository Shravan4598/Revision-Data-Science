````python
class NLPApp:
    def __init__(self):
        self.__database={}
        self.__first_menu()
    def __first_menu(self):
        first_input=input(
            """
            hii,
            Enter 0 to register:
            Enter 1 to login:
            Else to exit:
            """)
        if first_input=="0":
            self.__register()
        elif first_input=="1":
            self.__login()
        else:
            exit()
    def __register(self):
        email=input("Enter Email ID: ")
        name=input("Enter Name: ")
        password=input("Enter Password: ")
        if email in self.__database:
            print("Email is already exist, Login")
            self.__login()
        else:
            self.__database[email]=[name,password]
            print("Registration Successful. Now, You can Login.")
            self.__login()
    def __login(self):
        email=input("Enter your email id: ")
        password=input("Enter your passwrod: ")
        if email in self.__database:
            if self.__database[email][1]==password:
                print("Login successful")
                # self.__second_menu()
            else:
                print("Wrong Password. Try Again")
                self.__login()
        else:
            print("Your email is not registered. Please Register your email first")
            self.__register()
                
````
