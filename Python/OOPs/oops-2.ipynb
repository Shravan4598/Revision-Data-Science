# Reference Variable

```python
class Person:
    def __init__(self):
        self.name="Shravan"
        self.gender="male"
```

```python
Person()
```

- **Object will create**

```python
p=Person()
p
```

- **p is a varibale name which is reference variable of class and store the address of object created**

```python
p=Person()
q=p
print(id(p))
print(id(q))
print("id(p)==id(q):",id(p)==id(q))
```

```python
print(p.name)
print(q.name)
```

```python
q.name="Kumar"
print(p.name)
print(q.name)
```
