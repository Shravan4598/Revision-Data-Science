# Encapsulation
- Wrapping up data and methods

````python
class Person:
    def __init__(self,name):
        self.name=name
````

````python
p=Person("Shravan")
print(p.name)
````

````python
p.name="Kumar"
print(p.name)
````

- Here you can see name has been changed
- To overcome this issue we can make it private using (__)

````python
class Person:
    def __init__(self,name):
        self.__name=name
````

````python
p=Person("Shravan")
````

##### print(p.name)----> it will give error

- To access this we need a getter method to see the name but we cannot access it directly
- if we will access it through error

````python
class Person:
    def __init__(self,name):
        self.__name=name
    def get_name(self):
        return self.__name
````

````python
p=Person("Shravan")
p.get_name()
````

- We cannot change it directly.
- To change it we need another method called **setter method**

````python
class Person:
    def __init__(self,name):
        self.__name=name
    def get_name(self):
        return self.__name
    def set_name(self, new_name):
        self.__name = new_name
````

````python
p=Person("Shravan")
p.get_name()
````

````python
p.set_name("Kumar")
p.get_name()
````

- **As you can see it is changed with the help of setter**

````python
class Person:
    country = "India"

    def __init__(self, name):
        self.name = name

    @staticmethod
    def greet():
        print("Welcome to the Person class!")
````

````python
class Person:

    @staticmethod
    def is_adult(age):
        return age >= 18
````

# Instance Method vs Static Method

````python
class Person:
    country = "India"

    def __init__(self, name):
        self.name = name

    def display(self):           # Instance Method
        print(self.name)

    @staticmethod
    def greet():                 # Static Method
        print("Hello Everyone")
````

````python
p1 = Person("Shravan")

p1.display()
Person.greet()
````

```
Person Class
──────────────────────────

country = "India"     ← Static/Class Variable

display(self)         ← Instance Method
      │
      └── Needs object

greet()               ← Static Method
      │
      └── Needs neither object nor class
```

````python
class Person:
    country = "India"  # Static/Class Variable

    def __init__(self, name):
        self.name = name

    def display(self):
        print(f"Name: {self.name}")
        print(f"Country: {Person.country}")

    @staticmethod
    def greet():
        print("Welcome!")

    @staticmethod
    def is_adult(age):
        return age >= 18
````

````python
p1 = Person("Shravan")
p2 = Person("Rahul")

p1.display()
p2.display()

Person.greet()

print(Person.is_adult(20))
print(Person.is_adult(15))
````
