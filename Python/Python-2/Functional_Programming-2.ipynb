# Functional Programming

- map(function,iterable)
- map(lambda x:x*x,list1)
- return map object
- so we need to convert it to list
- list(map(lambda x:x*x,list1))

```python
A=[1,1,1,0,1,0,0]
B=[0,1,1,0,0,1,0]

# Return -> [False,True,True,True,False,False, True]
# Code
match=[]
for i in range(len(A)):
    if A[i]==B[i]:
        match.append(True)
    else:
        match.append(False)

print(match)
```

**Solving the above problem using map function**

```python
list1=[1,2,3,4]
list(map(lambda x:x*x,list1))
```

```python
list(map(lambda x,y: x==y,A,B))
```

```python
def sq_num(x):
    return x*x
list1=[1,2,3,4]
list(map(sq_num,list1))
```

# Filter
- filter out elements fromthe iterable based on condition
- filter(function,iterable)

```python
list2=[100,101,105,87]
list(filter(lambda x:x%2!=0,list2))
```

```python
even_num=[]
for i in list2:
    if i%2==0:
        even_num.append(i)
print(even_num)
```

- The above for loop code can be converted into one line by using filter function.

```python
list(filter(lambda x: x%2==0,list2))
```

- Syntax: reduce(function,iterable)

# Args and Kwargs
- **Args**: arbitrary postional arguments
- **kwargs**: keyword arbitrary arguments
- args stores value as  tuple
- kwargs strores value as  dictionary

```python
# Summasation function
def summation(a,b):
    return a+b
```

- This function can take only two number as input
- But in real we need many input for example calculator
- To make the function more flexible we need **args** and **kargs**

```python
def summation(a,b,*args):
    print("a:",a)
    print("b:",b)
    print("args:",args)
    return a+b
```

```python
summation(10,15)
```

```python
summation(10,15,18)
```

```python
summation(10,15,18,50,60,80)
```

```python
def summation(a,b,*args):
    print("a:",a)
    print("b:",b)
    result=a+b
    if args:
        result+=sum(args)
    print("args:",args)
    return result
```

```python
summation(10,15,18,50,60,80)
```

## Kwargs
- We are making a website in which there is a sinup option
- In sinup option, some fields are mandatory and some fields are not mandatory
- To fill the non mandatory fields we required **kwargs**

```python
def create_profile(name,gender,age):
    Person={"name":name,"Gender":gender,"Age":age}
    return Person
```

```python
create_profile(name="A",gender="Male",age=21)
```

- name, gender, age are mandatory fields if person provides any other information which is not mandatory
- Because company do not want to lose data regarding user
- Data is the new oil
- To handle this situation we need keyword argument,

```python
def create_profile(name,gender,age,**kwargs):
    Person={"name":name,"Gender":gender,"Age":age}
    return Person
```

```python
create_profile(name="A",gender="Male",age=21,hobby="Reading Book")
```

```python
def create_profile(name,gender,age,**kwargs):
    Person={"name":name,"Gender":gender,"Age":age}
    print(kwargs)
    return Person
```

```python
create_profile(name="A",gender="Male",age=21,hobby="Reading Book")
```

- Kwargs return dictionary

```python
def create_profile(name,gender,age,**kwargs):
    Person={"name":name,"Gender":gender,"Age":age}
    if kwargs:
        Person.update(kwargs)
    return Person
```

```python
create_profile(name="A",gender="Male",age=21,hobby="Reading Book")
```

```python
def fun(x,y,*args,**kwargs):
    print("x",x)
    print("y",y)
    print("args",args)
    print("kwargs",kwargs)
```

```python
fun(1,2,3,4)
```
