### Insights
- 4 state updates
- state update can be costly
- Time Complexity: O(n)
- Space Complexity: O(n)

### Solution: Map Function

### map(function,iterable)
- map applies the function on each element of the iterable
- return a map object
- lazy evaluation: compute the values only when it is aked for explicitly
- They avoid mutable state
- They are allowing optimization like memoization and safe parallelism which reduces redundatnt computations
- Example: map(lambda function,iterable)

```python
num=[1,2,3,4]
map(lambda x:x*x,num)
```

```python
list(map(lambda x:x*x,num))
```

```python
tuple(map(lambda x:x*x,num))
```

```python
set(map(lambda x:x*x,num))
```

## Higher Order Function(HOF)
- A function that can take another function as an argument or return a function (or both).
- 

```python
def fun():
    return 100
```

```python
fun()
```

```python
def outer():
    def inner():
        return 100
    return inner
```

```python
var1=outer()
var1
```

```python
var1() # Calling the inner function-->100
```

# Decorators

```python
def class_greet():
    print("Hii Guys")
```

```python
class_greet()
```

```python
def preety(x):
    def foo():
        print("-"*10)
        x()
        print("-"*10)
    return foo
```

```python
@preety
def class_greet():
    print("Hii Guys")
```

```python
class_greet()
```

```python
@preety
def hello():
    print("Hello World")
```

```python
hello()
```

### Decorators are built on the top of Higher Order  Function

### Decorators are HOF that take another function as input and adds some extra behaviour along with the functionality of the passed function

```python
from datetime import datetime
def runtime(x):
    def inner(p):
        start=datetime.now()
        print(x)
        print(p)
        num=x(p)
        end=datetime.now()
        print("Time to execute the function is ",end-start)
        return num
    return inner
```

```python
def square(num):
    return num**2
```

```python
square(5)
```

```python
@runtime
def square(num):
    return num**2
```

```python
square(5)
```

```python
list(map(square,[1,2,3,4]))
```
